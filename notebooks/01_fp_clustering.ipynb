# FP Clustering Demo

Run the scripts first, then use this notebook to inspect clusters.


```python
import pandas as pd
df = pd.read_csv('../data/processed/fp_clusters.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
plt.scatter(df['umap_x'], df['umap_y'], c=df['cluster_id'])
plt.title('False positive clusters')
plt.xlabel('UMAP x')
plt.ylabel('UMAP y')
plt.show()
```

```python
df.groupby(['cluster_id', 'synthetic_fp_type']).size().sort_values(ascending=False)
```
